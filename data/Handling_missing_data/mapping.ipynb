```python
import pandas as pd

# Load datasets
openpom_df = pd.read_csv("C:/Users/suman/OneDrive/Bureau/Internship_Study/GNN_On_OdorPrediction/data/openPOMData/curated_GS_LF_merged_4983.csv")
sos_df = pd.read_csv('C:/Users/suman/OneDrive/Bureau/Internship_Study/GNN_On_OdorPrediction/data/Data_Sampling/Saturated_SoS_Top138.csv',encoding='ISO-8859-1')

# Odor mapping: OpenPOM column (English) → SoS_Top138 column (French)
odor_mapping = {
    "grassy": "grass",
    "popcorn": "popcorn",
    "ozone": "ozone",
    "gassy": "gaz",               
    "dry": "sec",        
    "chamomile": "camomille",     # Not present — will be added
    "brandy": "brandy",           # Not present — will be added
    "black currant": "cassis",     # Not present — will be added
    "warm": "chaud",
    "terpenic": "terpenique",
    "sour": "acide",
    "sharp": "sharp",
    "savory": "savory",
    "pungent": "pungent",
    "orangeflower": "fleur_d_oranger",
    "natural": "natural",
    "meaty": "viande_cuite",
    "malty": "malte",
    "ketonic": "cétonique",
    "juicy": "juteux",
    "fruit skin": "fruit_a_pepins",  # Closest match in French
    "cortex": "cortex",
    "cooling": "cooling",
    "clean": "propre",
    "beefy": "viande_cuite"  # Maps to same as "meaty"
}

# Ensure 'smiles' column exists
if 'SMILES' not in openpom_df.columns or 'SMILES' not in sos_df.columns:
    raise ValueError("Both datasets must have a 'SMILES' column.")

# Process each odor mapping
for en_odor, fr_odor in odor_mapping.items():
    if en_odor not in openpom_df.columns:
        print(f"⚠️ OpenPOM odor '{en_odor}' not found. Skipping...")
        continue

    # Get all SMILES from OpenPOM with this odor
    matching_smiles = openpom_df.loc[openpom_df[en_odor] == 1, 'SMILES'].dropna().unique()
    
    # Add column if it doesn't exist in SoS_Top138
    if fr_odor not in sos_df.columns:
        sos_df[fr_odor] = 0  # Create column and initialize with 0
        print(f"Added new column '{fr_odor}' to SoS_Top138.")

    # Set 1 where SMILES match
    sos_df.loc[sos_df['SMILES'].isin(matching_smiles), fr_odor] = 1
    print(f"✓ Mapped OpenPOM '{en_odor}' → SoS_Top138 '{fr_odor}' ({len(matching_smiles)} SMILES updated)")

# Save result
sos_df.to_csv("SoS_Top138_updated.csv", index=False)
print("\nUpdated SoS_Top138 saved as 'SoS_Top138_updated.csv'")
```
