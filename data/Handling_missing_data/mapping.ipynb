```python
import csv
import pandas as pd
```

### Odor Mapping OpenPOM to SoS

```python
# Step 1: Load TGSC to SOS mapping
mapping = {}
sos_values = set()

mapping_file_path = r'C:/Users/suman/OneDrive/Bureau/Internship_Study/GNN_On_OdorPrediction/data/Handling_missing_data/croisement_sos_tgsc.csv'

with open(mapping_file_path, 'r', encoding='utf-8') as f:
    for line in f:
        parts = line.strip().split('\t')
        if len(parts) >= 2:
            tgsc = parts[0]
            sos = parts[1].strip()
            mapping[tgsc] = sos
            sos_values.add(sos)

print(f"Loaded {len(mapping)} mappings from TGSC to SOS.")
print(f"Found {len(sos_values)} unique SOS odor terms.\n")

# Step 2: Process OpenPOM dataset
input_file_path = r'C:/Users/suman/OneDrive/Bureau/Internship_Study/GNN_On_OdorPrediction/data/openPOMData/OpenPOM138.csv'
output_file_path = 'mapped_OpenPOM.csv'

unmapped_columns = set()
mapped_columns = set()
mapped_conversions = []  # (openpom odor, sos odor)

with open(input_file_path, 'r', encoding='utf-8') as infile, \
    open(output_file_path, 'w', newline='', encoding='utf-8') as outfile:

    reader = csv.reader(infile)
    headers = next(reader)

    # First two columns: SMILES and descriptors – leave unchanged
    fixed_headers = headers[:2]

    # Remaining columns: the odor columns
    odor_columns = headers[2:]

    for col in odor_columns:
        mapped = mapping.get(col, None)

        if mapped is not None:
            fixed_headers.append(mapped)
            mapped_columns.add(col)
            mapped_conversions.append((col, mapped))  # mapped via dictionary
        elif col in sos_values:
            fixed_headers.append(col)
            mapped_columns.add(col)
            mapped_conversions.append((col, col))     # already valid SOS
        else:
            print(f"Unmapped odor column: '{col}'")
            unmapped_columns.add(col)
            fixed_headers.append(col)

    # Write updated header
    writer = csv.writer(outfile)
    writer.writerow(fixed_headers)

    # Copy all data rows unchanged
    for row in reader:
        writer.writerow(row)

# Final Reporting
print(f"\nTotal odor columns: {len(odor_columns)}")
print(f"Number of mapped odor columns: {len(mapped_columns)}")
print(f"Number of unmapped odor columns: {len(unmapped_columns)}")

# Print mapped conversions (all of them)
print("\nMapped OpenPOM → SOS Odor Conversions (All):")
for i, (openpom, sos) in enumerate(sorted(mapped_conversions), 1):
    arrow = "➜" if openpom != sos else " →"
    print(f"{i:3}. {openpom} {arrow} {sos}")

# Print unmapped columns
print("\nUnmapped odor columns:")
for i, col in enumerate(sorted(unmapped_columns), 1):
    print(f"{i:3}. {col}")

print(f"\nMapped dataset saved to: {output_file_path}")
```
